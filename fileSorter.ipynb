**File sorter in Python**

```python
import os, shutil
```

To use it, replace the path given below with the path to the folder where the files are kept

```python
path = r"D:/DA/File sorter/"
```

```python
file_names = os.listdir(path)
```

in the folders array, list out all the types of folders you want for your files. in my case, i had a word file, csv file, and 2 images. hence i created text, database and image folders for them. 

```python
folders = ['databases', 'images', 'text files'] 

for loop in range(0,3):
    if  not os.path.exists(path+ folders[loop]):
        print(path+ folders[loop])
        os.makedirs(path+ folders[loop])
```

the following code snippet will sort your files for you

```python
for file in file_names: 
    if ('.sql' in file or '.csv' in file or '.xlsx' in file) and not os.path.exists(path+ 'databases/'+ file):
        shutil.move( path+ file ,path+ 'databases/'+ file)
    elif ('.txt' in file or '.docx' in file) and not os.path.exists(path+ 'text files/'+ file):
        shutil.move( path+ file ,path+ 'text files/'+ file)
    elif ('.png' in file or '.jpg' in file or '.jpeg' in file) and not os.path.exists(path+ 'images/'+ file):
        shutil.move( path+ file ,path+ 'images/'+ file)
    
```
